###  1--> Build a function that recursively prints the type of every value in a nested dictionary/list structure.

```python
data = {
    "Name": "Navi",
    "Age": 25,
    "Hobbies": ["Reading", "Gaming", {"Sports": ["Badminton", "Basketball"]}],
    "Education": {
        "School": {
            "Name": "Nachieketa  School"
        },
        "Undergraduate": {
            "Degree": "Bachelor of computer Application",
            "University": "Sage University"
        }
    },
    "Interests": ["Technology", "Music", {"Languages": ["Python", "JavaScript"]}]
}
```

```python
def print_type(data):
    if isinstance(data,dict):
        for key , value in data.items():
            print(f"Key {key}")
            print_type(value)

    elif isinstance(data,list):
        for index , value in enumerate(data):
            print(f"index {index}")
            print_type(value)
    else:
        print(f"value {data} --> type : {type(data).__name__}")


print_type(data)
```

### 2--> Implement a safe parser that converts strings to int, float, bool, or leaves them as str based on their content.


```python
String = input("Enter the string: ")

try:
    value = int(String)
    print(value)
    print(f"Type : {type(value).__name__}")

except ValueError:
    try:
        value = float(String)
        print(value)
        print(f"Type : {type(value).__name__}")

    except ValueError:
        if String.lower() == "true":
            value = True
            print(value)
            print(f"Type : {type(value).__name__}")

        elif String.lower() == "false":
            value = False
            print(value)
            print(f"Type : {type(value).__name__}")

        else:   
            print(String)
            print(f"Type : {type(String).__name__}")
```

### 3--> Calculate electricity bill

### unit consumed = [0-100 , 101-200 , Above 200]
### Rate per unit = [₹ 5 , ₹ 7 , ₹ 10]

```python
number_of_units = int(input("Enter the number of units consumed: "))
if number_of_units <= 100:
    electricity_bill = number_of_units * 5
elif number_of_units <= 200:
    electricity_bill = (100 * 5) + ((number_of_units - 100) * 7)
else:
    electricity_bill = (100 * 5) + (100 * 7) + ((number_of_units - 200) * 10)
print(f"ELectrivity Bill is : ₹ {electricity_bill}")
```

### 2 --> Build BMI calculator.

```python
print("=" * 50)
print("        BMI CALCULATOR")
print("=" * 50)

while True:

    try:
        name = input("\nEnter your name: ")

        age = int(input("Enter your age: "))

        gender = input("Enter your gender (Male/Female): ").strip().lower()

        if gender not in ["male", "female"]:
            print("Invalid gender! Please enter Male or Female.")
            continue

        if age < 18:
            print("\nBMI classification for people below 18 years")
            print("requires age and gender percentile charts.")
            print("This calculator supports only adults (18+).")
            continue

        print("\nHeight Unit")
        print("1. Centimeter")
        print("2. Meter")

        choice = int(input("Choose height unit (1 or 2): "))

        if choice == 1:
            height_cm = float(input("Enter height (cm): "))
            height = height_cm / 100

        elif choice == 2:
            height = float(input("Enter height (meter): "))

        else:
            print("Invalid Choice.")
            continue

        weight = float(input("Enter weight (kg): "))

        if height <= 0 or weight <= 0:
            print("Height and Weight must be greater than zero.")
            continue

        bmi = weight / (height ** 2)

        if bmi < 18.5:
            category = "Underweight"
            advice = "Eat a balanced diet and consult a nutritionist if required."

        elif bmi < 25:
            category = "Normal Weight"
            advice = "Great! Maintain your healthy lifestyle."

        elif bmi < 30:
            category = "Overweight"
            advice = "Regular exercise and a balanced diet are recommended."

        else:
            category = "Obese"
            advice = "Consult a healthcare professional for proper guidance."

        print("\n" + "=" * 50)
        print("              BMI REPORT")
        print("=" * 50)

        print(f"Name      : {name}")
        print(f"Age       : {age}")
        print(f"Gender    : {gender.capitalize()}")
        print(f"Height    : {height:.2f} meter")
        print(f"Weight    : {weight:.2f} kg")
        print(f"BMI       : {bmi:.2f}")
        print(f"Category  : {category}")

        print("\nHealth Advice:")
        print(advice)

        print("=" * 50)

        again = input("\nDo you want to calculate again? (yes/no): ").lower()

        if again != "yes":
            print("\nThank you for using BMI Calculator.")
            break

    except ValueError:
        print("Invalid Input! Please enter numbers correctly.")
```

### 3--> Build a grading calculator using comparison operators.

```python
marks = float(input("Enter your marks: "))

if marks < 0 or marks > 100:
    print("Invalid Marks!")

elif marks >= 90:
    print("Grade: A+")

elif marks >= 80:
    print("Grade: A")

elif marks >= 70:
    print("Grade: B")

elif marks >= 60:
    print("Grade: C")

elif marks >= 50:
    print("Grade: D")

else:
    print("Grade: Fail")
```
