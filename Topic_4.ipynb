#### 3--> Check whether a number is prime.

```python
number = int(input("Enter the number :"))


if number <= 1:
    print(f"{number} is not a prime number")
    exit()
else:
    Prime_number = True

for i in range(2, int(number ** 0.5) + 1):
    if number % i == 0:
         Prime_number = False
         break

if Prime_number:
    print(f"{number} is a prime number")
else:
    print(f"{number} is not a prime number")




```

#### 5--> Count the frequency of each character in a string.

```python
text = input("Enter the string: ")

freq = {}

for ch in text:
    if ch in freq:
        freq[ch] += 1
    else:
        freq[ch] = 1

print(freq)
```

#### 1--> Flatten a nested list using loops.

```python
person = ["name","Age",["gaming","cricket","badminton"],"good in study"]

def flatten(lst):
    result = []
    for i in lst:
        if isinstance(i, list):
            result.extend(flatten(i))
        else:
            result.append(i)
    return result

F = flatten(person)
print(F)
```

#### 2--> Simulate an ATM menu using a while loop.

```python
balance = 10000  

while True:
    print("\n===== ATM MENU =====")
    print("1. Check Balance")
    print("2. Deposit Money")
    print("3. Withdraw Money")
    print("4. Exit")

    choice = int(input("Enter your choice (1-4): "))

    if choice == 1:
        print(f"Your current balance is ₹{balance}")

    elif choice == 2:
        amount = float(input("Enter amount to deposit: ₹"))
        if amount > 0:
            balance += amount
            print(f"₹{amount} deposited successfully.")
            print(f"Updated Balance: ₹{balance}")
        else:
            print("Please enter a valid amount.")

    elif choice == 3:
        amount = float(input("Enter amount to withdraw: ₹"))
        if amount <= 0:
            print("Please enter a valid amount.")
        elif amount > balance:
            print("Insufficient balance!")
        else:
            balance -= amount
            print(f"₹{amount} withdrawn successfully.")
            print(f"Remaining Balance: ₹{balance}")

    elif choice == 4:
        print("Thank you for using the ATM!")
        break

    else:
        print("Invalid choice! Please select between 1 and 4.")
```

#### 4--> Read a CSV file and compute average sales per region without using Pandas.

```python
file_read = open("sales_data (1).csv", "r")
file_content = file_read.readlines()
print(file_content)
file_read.close()
```

```python
newdata = []
for line in file_content:
    newdata.append(line.strip().split(','))
print(newdata)
```

```python
region_wise_sales = {}
region_Count = {}

for row in newdata[1:]:
    region = row[0]
    sales = int(row[1])

    if region in region_wise_sales:
        region_wise_sales[region] +=  sales
        region_Count[region] += 1
    else:
        region_wise_sales[region] = sales
        region_Count[region] = 1

print(region_wise_sales)
print(region_Count)
```

```python
for region in region_wise_sales:
    average_sales = region_wise_sales[region] / region_Count[region]
    print(f"Region: {region}, Total Sales: {region_wise_sales[region]}, Count: {region_Count[region]}, Average Sales: {average_sales:.2f}")
```

#### 5--> Detect duplicate values in a list without using set().

```python
list1 = [1,2,6,3,4,6]
for i in range(len(list1)-1):
    for j in range(i+1, len(list1)):
        if list1[i] == list1[j]:
            print(f"Duplicate found: {list1[j]} at index {j}")
```
